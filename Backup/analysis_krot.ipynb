```python
import numpy as np
import scipy
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
from joblib import Parallel, delayed
import lattice
import importlib
import kmc
import loading as ld
importlib.reload(kmc)
```

```python
# 1D

F=1
tau = 1
r0= 1
n= 50
L =  r0*np.array([n])
PBC = [0]
LOAD_FIXED_KAPPA= False
FIXED_KAPPA = 0.476#2/3
save = False
cutoff = r0


nrounds = 3000

KF = (RF/r0)**6/tau
krot_dip= np.linspace(10,1000,3)*KF
kappas2_LD = []
for i in range(len(krot_dip)):
    krot_i = krot_dip[i]
    syst = ld.System(save=save,RF=RF,tau=tau,r0=r0,cutoff=cutoff,L=L,LOAD_FIXED_KAPPA=LOAD_FIXED_KAPPA,FIXED_KAPPA=FIXED_KAPPA,n=n,PBC=PBC,krot_dip=krot_i)
    syst.make_lattice()
    
    results = Parallel(n_jobs=6,verbose=10)(delayed(kmc.kmc_run_fluor)(syst) for n in range(nrounds))
    out_dict = kmc.process_results(results,file_header="",dim=len(L),RF=RF,tau=tau,r0=r0,FIXED_KAPPA=FIXED_KAPPA)
    kappas2_LD.append(out_dict['kappa2_LD'])

np.savetxt("kappas2_Krot_1D_1.txt", kappas2_LD)
```
