```python
latt = np.loadtxt("lattice.txt")
x = latt[:,0]
y = latt[:,1]
z = latt[:,2]

dx = x.max() - x.min()
dy = y.max() - y.min()
dz = z.max() - z.min()

dists_x = []
dists_y = []
dists_z = []
for i in range(len(x)):
   for j in range(len(x)):
      if i != j:
        dist_x = x[i]-x[j]
        dist_y = y[i]-y[j]
        dist_z = z[i]-z[j]
        dists_x.append(dist_x)
        dists_y.append(dist_y)
        dists_z.append(dist_z)

dist_x_mean = np.mean(dists_x)
dist_y_mean = np.mean(dists_y)
dist_z_mean = np.mean(dists_z)

L = np.array([dx+dist_x_mean, dy+dist_y_mean, dz+dist_z_mean])
L = np.array([300,300,300])
```

```python
# Dados IT4F


RF= 26.5/(0.476)**(1/3)
tau = 351
r0= 20
n = 1000
PBC = [0,1]
LOAD_FIXED_KAPPA= False
FIXED_KAPPA = 0.476#2/3
save = False
cutoff = 200
nrounds = 10000
krot_dip = 0 #=0 -> no rotation

latt = np.loadtxt("lattice.txt")
X,Y,Z = latt[:,0],latt[:,1],latt[:,2]

dist_X = np.mean(X)
dist_Y = np.mean(Y)
dist_Z = np.mean(Z)

dX = X.max()-X.min() + dist_X
dY = Y.max()-Y.min() + dist_Y
dZ = Z.max()-Z.min() + dist_Z

L = np.array([dX,dY,dZ])

#detail: Only one syst object, so all dynamics here have the same lattice and orientational profile
syst = ld.System(save=save,RF=RF,tau=tau,r0=r0,cutoff=cutoff,L=L,LOAD_FIXED_KAPPA=LOAD_FIXED_KAPPA,FIXED_KAPPA=FIXED_KAPPA,n=n,PBC=PBC,krot_dip=krot_dip)
syst.make_lattice(loadcoord=True,loaddip=True)
results = Parallel(n_jobs=10,verbose=10)(delayed(kmc.kmc_run_fluor)(syst) for n in range(nrounds))



out_dict = kmc.process_results(results,file_header="",dim=len(L),RF=RF,tau=tau,r0=r0,FIXED_KAPPA=FIXED_KAPPA)
```
